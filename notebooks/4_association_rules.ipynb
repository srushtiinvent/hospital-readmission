```python
import pandas as pd
import numpy as np

# Load data
df = pd.read_csv('../data/cleaned_data.csv')

# Identify readmission patterns
readmitted = df[df['readmitted'] == 1]
not_readmitted = df[df['readmitted'] == 0]

print(f"Total patients: {len(df)}")
print(f"Readmitted patients: {len(readmitted)} ({len(readmitted)/len(df)*100:.1f}%)")
print(f"Not readmitted: {len(not_readmitted)} ({len(not_readmitted)/len(df)*100:.1f}%)")
```
